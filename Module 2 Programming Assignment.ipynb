# 6.2
In this Excercise I decided to have the number increment up if it was too low, and increment down if it was too high, because the assignment description had me with a loop that would have just kept printing "too low" indefinitely.

```python
guess_me = 7
```

```python
number = 1
```

```python
while True:
    if(number < guess_me):
        print("too low")
        number += 1
    elif(number == guess_me):
        print("found it!")
        break
    else:
        print("oops.")
        number -= 1
```
